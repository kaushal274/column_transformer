```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(df.drop(columns=['customer_id','age','churn','city','job','education']),df['churn'],test_size=0.2,)
```

```python
X_train.isnull().sum()
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
transformer=ColumnTransformer([
    ('tnf1',SimpleImputer(strategy='median'),['years_experience','annual_income','credit_score','avg_transaction_value']),
    ('tnf3',OneHotEncoder(sparse_output=False,drop='first',handle_unknown='ignore'),['gender','marital_status','has_credit_card','channel','segment'])
],remainder='drop')
```

```python
transformer.fit_transform(X_train)
transformer.transform(X_test)
```
